## **Question 1:** What is React?

### **Answer**

**React** is a free, open-source JavaScript library (created and maintained by Meta)
used for building fast, interactive user interfaces, especially single-page applications.

Key characteristics:
- **Component-based:** UIs are built from small, reusable, self-contained pieces called
  *components*, each managing its own structure, logic, and appearance.
- **Declarative:** Developers describe *what* the UI should look like for a given state,
  and React handles updating the actual DOM to match — instead of manually manipulating
  the DOM step by step.
- **Virtual DOM:** React maintains an in-memory representation of the UI (the Virtual
  DOM). When state changes, React efficiently compares (diffs) the new Virtual DOM with
  the previous one and updates only the parts of the real DOM that actually changed,
  improving performance.
- **Unidirectional (one-way) data flow:** Data flows from parent components down to
  child components via `props`, making applications easier to understand and debug.
- **Reusable & composable:** Components can be nested, reused, and composed together to
  build complex UIs from simple building blocks.


## **Question 2:** Write a React functional component named `Greeting` that accepts a `name` prop and displays: `"Hello, [name]!"`.

### **Answer**

```jsx
function Greeting({ name }) {
  return <h2>Hello, {name}!</h2>;
}

export default Greeting;
```

**Usage:**
```jsx
<Greeting name="Alice" />
// Renders: Hello, Alice!
```

**Explanation:**
- `Greeting` is a **functional component** — a plain JavaScript function that returns
  JSX.
- `{ name }` destructures the `name` value directly out of the `props` object.
- `{name}` inside the JSX interpolates the value of the `name` prop into the rendered
  output.


## **Question 3:** What is JSX and why is it useful in React?

### **Answer**

**JSX (JavaScript XML)** is a syntax extension for JavaScript that allows you to write
HTML-like markup directly inside JavaScript code. For example:

```jsx
const element = <h1>Hello, world!</h1>;
```

Behind the scenes, JSX is not understood by browsers directly — tools like Babel
compile it into regular JavaScript function calls, such as:

```javascript
const element = React.createElement('h1', null, 'Hello, world!');
```

**Why JSX is useful:**
- **Readability:** It lets developers describe UI structure in a familiar, HTML-like
  syntax rather than deeply nested `createElement` calls, making code easier to read and
  write.
- **Combines logic and markup:** JSX allows embedding JavaScript expressions directly
  within markup using curly braces `{ }`, keeping rendering logic and UI structure
  together in one place.
- **Catches errors early:** Because JSX compiles to JavaScript, syntax and reference
  errors are often caught at compile time rather than at runtime.
- **Component composition:** JSX makes it natural to nest and compose components (e.g.
  `<Header /><Content /><Footer />`), which mirrors how React applications are
  structured.


## **Question 4:** Explain how props support React's one-way data flow.

### **Answer**

**Props (short for "properties")** are read-only inputs passed from a parent component
to a child component. They enable **one-way (unidirectional) data flow**, meaning data
always flows *down* the component tree — from parent to child — and never automatically
flows back up.

```jsx
function Parent() {
  const message = "Hello from Parent";
  return <Child text={message} />;
}

function Child({ text }) {
  return <p>{text}</p>;
}
```

**How this supports one-way data flow:**
- A parent component owns and controls its data (state). It passes that data down to
  children as props.
- Child components **cannot modify** the props they receive directly — props are
  immutable from the child's perspective. If a child needs to change data, the parent
  must pass down a *callback function* as a prop, which the child calls to request a
  change (the parent then updates its own state).
- This one-directional flow makes applications more **predictable and easier to debug**,
  since data changes can always be traced back to a single source (the component that
  owns the state), rather than being modified from arbitrary places in the tree.


## **Question 5:** Create a functional component `ClickMeButton` that renders a button. It accepts `onClick` and `label` props. When clicked, it executes the passed `onClick` handler.

### **Answer**

```jsx
function ClickMeButton({ onClick, label }) {
  return (
    <button onClick={onClick}>
      {label}
    </button>
  );
}

export default ClickMeButton;
```

**Usage:**
```jsx
function App() {
  const handleClick = () => {
    console.log('Button was clicked!');
  };

  return <ClickMeButton onClick={handleClick} label="Click Me" />;
}
```

**Explanation:**
- `ClickMeButton` receives two props: `onClick` (a function) and `label` (a string).
- The `label` prop is rendered as the button's visible text.
- The `onClick` prop is attached directly to the native `onClick` event of the `<button>`
  element, so whenever the button is clicked, the function passed in by the parent
  component is executed.


### **Question 6:** Given `const fruits = ['Apple', 'Banana', 'Cherry']`, write a component `FruitList` that renders an unordered list of these fruits using `.map()` and includes a `key`.

### **Answer**

```jsx
function FruitList() {
  const fruits = ['Apple', 'Banana', 'Cherry'];

  return (
    <ul>
      {fruits.map((fruit, index) => (
        <li key={index}>{fruit}</li>
      ))}
    </ul>
  );
}

export default FruitList;
```

**Explanation:**
- `.map()` transforms each string in the `fruits` array into a `<li>` JSX element.
- Each `<li>` is given a unique `key` prop (here, the array `index`) so React can
  efficiently identify and track each list item across re-renders.
- The resulting array of `<li>` elements is placed inside a `<ul>` to render the full
  list.

*(Note: using the array index as a key is acceptable here since the list is static and
never reordered. For dynamic lists, a stable unique ID is preferred — see Question 7.)*


## **Question 7:** Why is the `key` attribute important when rendering lists in React? What goes wrong if keys are missing or misuse array index?

### **Answer**

**Why `key` is important:**
The `key` attribute gives each element in a list a **stable, unique identity**. React
uses these keys during its reconciliation (diffing) process to determine which items
have changed, been added, or been removed between renders — without keys, React would
have no reliable way to match old list items to new ones.

**What goes wrong without proper keys:**
- **Missing keys:** React falls back to using the array index by default and logs a
  console warning. This works fine only for lists that never change order or size.
- **Using array index as key (for dynamic lists):** If items are reordered, inserted, or
  removed, the index-to-item mapping shifts. React may then:
  - Incorrectly reuse DOM nodes for the wrong data, causing **stale or mismatched UI**
    (e.g. text updating on the wrong row).
  - Cause **component state to "jump"** between items (e.g. an input's typed text
    appearing on the wrong row after reordering).
  - Lead to **unnecessary re-renders** or **loss of performance benefits**, since React
    can't correctly identify which elements are truly unchanged.
- **Best practice:** Always use a stable, unique identifier from the data itself (like a
  database `id`) as the `key`, rather than the array index, whenever the list can be
  reordered, filtered, or modified.


## **Question 8:** Write a component `UserList` that receives a `users` prop — an array of objects each with `id` and `name`. Render each user's name in a `<p>` tag, using `map()` and `key`.

### **Answer**

```jsx
function UserList({ users }) {
  return (
    <div>
      {users.map((user) => (
        <p key={user.id}>{user.name}</p>
      ))}
    </div>
  );
}

export default UserList;
```

**Usage:**
```jsx
const users = [
  { id: 1, name: 'Alice' },
  { id: 2, name: 'Bob' },
  { id: 3, name: 'Charlie' },
];

<UserList users={users} />
```

**Explanation:**
- `UserList` destructures the `users` array directly from its props.
- `.map()` iterates over each user object and returns a `<p>` element containing that
  user's `name`.
- `key={user.id}` uses the unique `id` field from the data itself (rather than the array
  index), which is the recommended approach since user objects have a stable unique
  identifier.


## **Question 9:** Create an `App` component that holds an array of numbers, passes them to a `NumberList` child component for rendering, and passes a click handler function to log the clicked number. Show one-way data flow and event handling.

### **Answer**

```jsx
// NumberList.jsx
function NumberList({ numbers, onNumberClick }) {
  return (
    <ul>
      {numbers.map((num, index) => (
        <li key={index} onClick={() => onNumberClick(num)}>
          {num}
        </li>
      ))}
    </ul>
  );
}

export default NumberList;
```

```jsx
// App.jsx
import NumberList from './NumberList';

function App() {
  const numbers = [10, 20, 30, 40, 50];

  const handleNumberClick = (number) => {
    console.log('Clicked number:', number);
  };

  return (
    <div>
      <h2>Click a Number</h2>
      <NumberList numbers={numbers} onNumberClick={handleNumberClick} />
    </div>
  );
}

export default App;
```

**Explanation of one-way data flow & event handling:**
- **Data flows down:** `App` owns the `numbers` array (its data/state) and passes it
  *down* to `NumberList` as a prop — the child never modifies this array directly, only
  displays it.
- **Events flow up via callbacks:** `App` also passes a `handleNumberClick` *function*
  down as the `onNumberClick` prop. When a list item is clicked, `NumberList` doesn't log
  the number itself — it calls the function it received from the parent, effectively
  sending information "back up" through a callback rather than directly mutating parent
  state. This is the standard pattern React uses to combine one-way data flow with
  interactive event handling.


## **Question 10:** Design a component `ProductList` that receives `products` (array of objects with `id`, `name`, `price`). Render a list of product names and prices using `.map()`. Each item when clicked should call a prop `onProductSelect(product)`. Include the proper `key`.

### **Answer**

```jsx
function ProductList({ products, onProductSelect }) {
  return (
    <ul>
      {products.map((product) => (
        <li
          key={product.id}
          onClick={() => onProductSelect(product)}
          style={{ cursor: 'pointer' }}
        >
          {product.name} - ${product.price}
        </li>
      ))}
    </ul>
  );
}

export default ProductList;
```

**Usage:**
```jsx
function App() {
  const products = [
    { id: 1, name: 'Laptop', price: 999 },
    { id: 2, name: 'Phone', price: 599 },
    { id: 3, name: 'Headphones', price: 149 },
  ];

  const handleProductSelect = (product) => {
    console.log('Selected product:', product);
  };

  return <ProductList products={products} onProductSelect={handleProductSelect} />;
}
```

**Explanation:**
- `.map()` iterates over the `products` array, rendering each item's `name` and `price`
  inside an `<li>`.
- `key={product.id}` uses the unique `id` from each product object, which is the correct
  practice for dynamic, data-driven lists.
- Each `<li>` has an `onClick` handler that calls `onProductSelect(product)`, passing the
  **entire clicked product object** up to the parent component — demonstrating how child
  components communicate events (and associated data) back up via callback props.
