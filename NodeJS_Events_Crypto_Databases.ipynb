## **Question 1:** Create a custom event emitter that fires a `'userLoggedIn'` event with a username payload. Set up a listener to print a welcome message when the event occurs.

**Answer:**

```js
// events.js
const EventEmitter = require('events');

class MyEmitter extends EventEmitter {}

const myEmitter = new MyEmitter();

// Listener: runs whenever 'userLoggedIn' is emitted
myEmitter.on('userLoggedIn', (username) => {
  console.log(`Welcome, ${username}! You have successfully logged in.`);
});

// Fire the event with a username payload
myEmitter.emit('userLoggedIn', 'Alice');
```

**Output:**
```
Welcome, Alice! You have successfully logged in.
```

**Explanation:**
- Node's built-in `events` module provides the `EventEmitter` class, which implements the publish/subscribe pattern.
- `myEmitter.on('userLoggedIn', callback)` registers a **listener** for that event.
- `myEmitter.emit('userLoggedIn', 'Alice')` **fires** the event, passing `'Alice'` as the payload, which the listener receives as its `username` argument.

## **Question 2:** Use the `crypto` module to create an SHA256 hash for the string `"password123"`.

**Answer:**

```js
// hash.js
const crypto = require('crypto');

const hash = crypto.createHash('sha256').update('password123').digest('hex');

console.log(hash);
// e.g. ef92b778bafe771e89245b89ecbc08a44a4e166c06659911881f383d4473e94
```

**Explanation:**
- `crypto.createHash('sha256')` creates a hash object using the SHA256 algorithm.
- `.update('password123')` feeds the input data into the hash.
- `.digest('hex')` finalizes the hash and returns it as a hexadecimal string.

*(Note: hashing alone is not sufficient for storing real passwords — a proper password hashing algorithm like `bcrypt` or `argon2`, which adds salting and is deliberately slow, should be used in production instead of a plain SHA256 hash.)*

## **Question 3:** Use the `crypto` module to generate a secure, random Version 4 UUID.

**Answer:**

```js
// uuid.js
const crypto = require('crypto');

const id = crypto.randomUUID();

console.log(id);
// e.g. 3f9a1c2e-4b5d-4a6f-8e9a-1b2c3d4e5f6a
```

`crypto.randomUUID()` (available natively in Node.js 14.17+) generates a cryptographically secure, random Version 4 UUID — a 128-bit identifier commonly used for things like database primary keys, session IDs, or request tracing, without needing an external package.

## **Question 4:** Write a Node.js script that reads a `PORT` environment variable. If it's not set, it should default to `3000`.

**Answer:**

```js
// config.js
const PORT = process.env.PORT || 3000;

console.log(`Server will run on port: ${PORT}`);
```

`process.env` is an object containing all environment variables available to the Node.js process. `process.env.PORT` reads the `PORT` variable if it's set (e.g. exported in the shell or provided by a hosting platform); the `|| 3000` fallback ensures a sensible default is used if it's `undefined`.

**Run with a custom port:**
```bash
PORT=8080 node config.js
# Output: Server will run on port: 8080
```
**Run without setting PORT:**
```bash
node config.js
# Output: Server will run on port: 3000
```

## **Question 5:** Write the minimal code for an Express.js server that listens on port 3000 and responds to GET requests at the root URL (`/`) with "Hello, Express!".

**Answer:**

```js
// server.js
const express = require('express');
const app = express();

app.get('/', (req, res) => {
  res.send('Hello, Express!');
});

app.listen(3000, () => {
  console.log('Server listening on http://localhost:3000');
});
```

**Setup:**
```bash
npm init -y
npm install express
node server.js
```

`express()` creates an application instance. `app.get('/', handler)` registers a handler for GET requests to `/`. `res.send()` sends the response body and automatically sets the appropriate headers. `app.listen(3000, ...)` starts the server on port 3000.

## **Question 6:** Write a SQL statement to create a `users` table with an auto-incrementing primary key, an email, a password hash, and a creation timestamp.

**Answer:**

```sql
CREATE TABLE users (
  id INT AUTO_INCREMENT PRIMARY KEY,
  email VARCHAR(255) NOT NULL UNIQUE,
  password_hash VARCHAR(255) NOT NULL,
  created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);
```

**Explanation:**
- `id INT AUTO_INCREMENT PRIMARY KEY` creates a unique, auto-incrementing identifier for each row.
- `email VARCHAR(255) NOT NULL UNIQUE` stores the user's email, requiring it to be present and unique across all users.
- `password_hash VARCHAR(255) NOT NULL` stores the hashed password (never the plain-text password).
- `created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP` automatically records when each row was inserted.

*(This syntax targets MySQL; in PostgreSQL, `AUTO_INCREMENT` would be replaced with `SERIAL` or `GENERATED ALWAYS AS IDENTITY`.)*

## **Question 8:** Write a SQL query to select the `id` and `email` of the user whose email is "test@example.com".

**Answer:**

```sql
SELECT id, email
FROM users
WHERE email = 'test@example.com';
```

The `WHERE` clause filters rows to only the one matching the given email, and the `SELECT id, email` list restricts the result to just those two columns instead of returning the entire row.

## **Question 9:** Write a SQL statement to update the email of the user with an `id` of 1 to "new.email@example.com".

**Answer:**

```sql
UPDATE users
SET email = 'new.email@example.com'
WHERE id = 1;
```

`UPDATE ... SET ... WHERE ...` modifies existing rows. The `WHERE id = 1` clause is essential — without it, **every row** in the table would be updated to the same email.

## **Question 10:** Given a `users` table and a `posts` table (with a `user_id` foreign key), write a query to retrieve all posts and the email of the user who wrote them.

**Answer:**

```sql
SELECT posts.*, users.email
FROM posts
JOIN users ON posts.user_id = users.id;
```

**Explanation:**
- `posts.user_id` is a foreign key referencing `users.id`, linking each post to the user who created it.
- `JOIN users ON posts.user_id = users.id` (an inner join) combines each row in `posts` with the matching row in `users`.
- `SELECT posts.*, users.email` returns every column from `posts`, plus the `email` column pulled in from the joined `users` table — giving a combined result of "all post data + the author's email" in a single query.
