## **Question 1:** Using the `mysql2` library, write a Node.js script to establish a connection to a local MySQL database named `app_db`.

**Answer:**

```js
// db.js
const mysql = require('mysql2');

const connection = mysql.createConnection({
  host: 'localhost',
  user: 'root',
  password: 'your_password',
  database: 'app_db',
});

connection.connect((err) => {
  if (err) {
    console.error('Error connecting to the database:', err.message);
    return;
  }
  console.log('Connected to MySQL database: app_db');
});
```

**Setup:**
```bash
npm install mysql2
```

`mysql.createConnection()` takes the connection details (host, user, password, database name) and returns a connection object. `connection.connect()` actually opens the connection, running the given callback once it succeeds or fails.

## **Question 2:** Using an active `mysql2` connection, write a function to fetch all rows from a `products` table and log the results.

**Answer:**

```js
// getProducts.js
function getAllProducts(connection) {
  connection.query('SELECT * FROM products', (err, results) => {
    if (err) {
      console.error('Error fetching products:', err.message);
      return;
    }
    console.log('Products:', results);
  });
}

module.exports = getAllProducts;
```

`connection.query(sql, callback)` sends the SQL statement to MySQL and runs the callback once results come back. `results` is an array of row objects, each representing one row from the `products` table.

## **Question 3:** Define a Sequelize model for a `Task` that has a `title` (string) and a `completed` (boolean) field.

**Answer:**

```js
// models/Task.js
const { DataTypes } = require('sequelize');
const sequelize = require('../db'); // your configured Sequelize instance

const Task = sequelize.define('Task', {
  title: {
    type: DataTypes.STRING,
    allowNull: false,
  },
  completed: {
    type: DataTypes.BOOLEAN,
    defaultValue: false,
  },
});

module.exports = Task;
```

**Setup:**
```bash
npm install sequelize mysql2
```

`sequelize.define('Task', { ... })` creates a model mapped to a `Tasks` table (Sequelize pluralizes the model name by default). Each key in the object defines a column: `title` is a required string, and `completed` is a boolean that defaults to `false`.

## **Question 4:** Using a Sequelize model named `Task`, write a query to find all tasks and log their titles.

**Answer:**

```js
// listTasks.js
const Task = require('./models/Task');

async function listTaskTitles() {
  try {
    const tasks = await Task.findAll();
    tasks.forEach((task) => {
      console.log(task.title);
    });
  } catch (error) {
    console.error('Error fetching tasks:', error.message);
  }
}

listTaskTitles();
```

`Task.findAll()` returns a Promise that resolves to an array of `Task` model instances representing every row in the table. Each instance's properties (like `.title`) map directly to the corresponding table columns.

## **Question 6:** Write the code to connect to a local MongoDB server and a database named `inventory` using the official `mongodb` driver.

**Answer:**

```js
// db.js
const { MongoClient } = require('mongodb');

const uri = 'mongodb://localhost:27017';
const client = new MongoClient(uri);

async function connectDB() {
  try {
    await client.connect();
    console.log('Connected to MongoDB');
    const db = client.db('inventory');
    return db;
  } catch (error) {
    console.error('Error connecting to MongoDB:', error.message);
  }
}

module.exports = connectDB;
```

**Setup:**
```bash
npm install mongodb
```

`new MongoClient(uri)` creates a client pointed at the local MongoDB server. `client.connect()` opens the connection, and `client.db('inventory')` returns a reference to the `inventory` database, which can then be used to access its collections.

## **Question 7:** Using the native MongoDB driver, insert a new document into a `users` collection with a `name` of "John Doe" and an `age` of 30.

**Answer:**

```js
// insertUser.js
async function insertUser(db) {
  try {
    const result = await db.collection('users').insertOne({
      name: 'John Doe',
      age: 30,
    });
    console.log('Inserted user with _id:', result.insertedId);
  } catch (error) {
    console.error('Error inserting user:', error.message);
  }
}

module.exports = insertUser;
```

`db.collection('users')` gets a reference to the `users` collection. `.insertOne(document)` inserts a single document and returns a result object containing the newly generated `insertedId`.

## **Question 8:** Write a MongoDB query to find all users in a collection who are older than 25.

**Answer:**

```js
// findOlderUsers.js
async function findOlderUsers(db) {
  try {
    const users = await db.collection('users').find({ age: { $gt: 25 } }).toArray();
    console.log('Users older than 25:', users);
  } catch (error) {
    console.error('Error finding users:', error.message);
  }
}

module.exports = findOlderUsers;
```

`{ age: { $gt: 25 } }` is a MongoDB query filter using the `$gt` (greater than) operator to match documents where `age` is strictly greater than 25. `.find()` returns a cursor, and `.toArray()` converts all matching documents into a plain JavaScript array.

## **Question 9:** Write a MongoDB query to update a user's `age` to 31, finding the user by their `name`, "John Doe".

**Answer:**

```js
// updateUserAge.js
async function updateUserAge(db) {
  try {
    const result = await db.collection('users').updateOne(
      { name: 'John Doe' },
      { $set: { age: 31 } }
    );
    console.log('Matched:', result.matchedCount, 'Modified:', result.modifiedCount);
  } catch (error) {
    console.error('Error updating user:', error.message);
  }
}

module.exports = updateUserAge;
```

`.updateOne(filter, update)` finds the **first** document matching the filter (`{ name: 'John Doe' }`) and applies the update. `$set: { age: 31 }` updates only the `age` field, leaving every other field on the document untouched.

## **Question 10:** Write a MongoDB query to delete the user with the name "John Doe".

**Answer:**

```js
// deleteUser.js
async function deleteUser(db) {
  try {
    const result = await db.collection('users').deleteOne({ name: 'John Doe' });
    console.log('Deleted count:', result.deletedCount);
  } catch (error) {
    console.error('Error deleting user:', error.message);
  }
}

module.exports = deleteUser;
```

`.deleteOne(filter)` removes the **first** document matching the given filter (`{ name: 'John Doe' }`) from the collection. `result.deletedCount` confirms how many documents were actually removed (0 if no match was found, 1 if successful).
