## **Question 2:** Create a simple CommonJS module that exports a function to add two numbers. Then, create a separate script that imports and uses this function.

**Answer:**

```js
// math.js
function add(a, b) {
  return a + b;
}

module.exports = add;
```

```js
// app.js
const add = require('./math');

const result = add(5, 3);
console.log('5 + 3 =', result); // 5 + 3 = 8
```

`module.exports` marks what a file makes available to other files. `require('./math')` then imports that exported value in `app.js`, so `add` can be called directly. Run it with:
```bash
node app.js
```

## **Question 5:** Using the `fs/promises` module, write a script to asynchronously read the content of a file named `data.txt` and log it to the console.

**Answer:**

```js
// readFile.js
const fs = require('fs/promises');

async function readData() {
  try {
    const content = await fs.readFile('data.txt', 'utf-8');
    console.log(content);
  } catch (error) {
    console.error('Error reading file:', error.message);
  }
}

readData();
```

`fs/promises` is the Promise-based version of the `fs` module, so `readFile` can be `await`-ed inside an `async` function instead of using a callback. The `'utf-8'` encoding tells Node to return the file content as a readable string rather than a raw `Buffer`.

## **Question 6:** Using the `fs/promises` module, write a script to create a new file named `log.txt` and write the string "Task completed." into it.

**Answer:**

```js
// writeFile.js
const fs = require('fs/promises');

async function writeLog() {
  try {
    await fs.writeFile('log.txt', 'Task completed.');
    console.log('File written successfully!');
  } catch (error) {
    console.error('Error writing file:', error.message);
  }
}

writeLog();
```

`fs.writeFile(path, data)` creates the file if it doesn't already exist, or **overwrites** it if it does, then writes the given string into it. Using `await` lets us write this asynchronous operation in a clean, sequential style.

## **Question 7:** Use the `path` module to safely join the segments `src`, `utils`, and `helpers.js` into a single, cross-platform-compatible file path.

**Answer:**

```js
// pathDemo.js
const path = require('path');

const fullPath = path.join('src', 'utils', 'helpers.js');

console.log(fullPath);
// On Linux/Mac: src/utils/helpers.js
// On Windows:   src\utils\helpers.js
```

`path.join()` combines path segments using the **correct separator for the current operating system** (`/` on Linux/Mac, `\` on Windows), and also normalizes the result (removing redundant slashes, resolving `..` segments, etc.) — so the code works correctly no matter what platform it runs on.

## **Question 8:** Create a basic HTTP server using the `http` module that listens on port 3000 and responds to all requests with "Server is running."

**Answer:**

```js
// server.js
const http = require('http');

const server = http.createServer((req, res) => {
  res.statusCode = 200;
  res.setHeader('Content-Type', 'text/plain');
  res.end('Server is running.');
});

server.listen(3000, () => {
  console.log('Server listening on http://localhost:3000');
});
```

`http.createServer()` takes a callback that runs for **every incoming request**, receiving a `req` (request) and `res` (response) object. `server.listen(3000, ...)` starts the server on port 3000. Since this handler doesn't check the URL, every request — regardless of path — receives the same "Server is running." response.

## **Question 9:** Enhance the previous HTTP server to handle two different routes: `/` should respond with "Homepage", and `/about` should respond with "About Us".

**Answer:**

```js
// server.js
const http = require('http');

const server = http.createServer((req, res) => {
  res.setHeader('Content-Type', 'text/plain');

  if (req.url === '/') {
    res.statusCode = 200;
    res.end('Homepage');
  } else if (req.url === '/about') {
    res.statusCode = 200;
    res.end('About Us');
  } else {
    res.statusCode = 404;
    res.end('Not Found');
  }
});

server.listen(3000, () => {
  console.log('Server listening on http://localhost:3000');
});
```

`req.url` holds the path that was requested (e.g. `/` or `/about`). Checking it with `if`/`else if` lets the server respond differently based on the route — this simple pattern is essentially what routing frameworks like Express automate for larger applications. Any unmatched route falls through to a `404` response.

## **Question 10:** Write a script that efficiently reads a large file (`large.txt`) and outputs its content to the console using streams.

**Answer:**

```js
// streamRead.js
const fs = require('fs');

const readStream = fs.createReadStream('large.txt', { encoding: 'utf-8' });

readStream.on('data', (chunk) => {
  process.stdout.write(chunk);
});

readStream.on('end', () => {
  console.log('\n--- Finished reading file ---');
});

readStream.on('error', (error) => {
  console.error('Error reading file:', error.message);
});
```

**Explanation:**
- `fs.createReadStream()` reads the file in small **chunks** instead of loading the entire file into memory at once (unlike `fs.readFile`), which makes it far more memory-efficient for large files.
- The `'data'` event fires each time a new chunk is read; `process.stdout.write(chunk)` prints each chunk as it arrives, so output starts immediately instead of waiting for the whole file to be read.
- The `'end'` event fires once all data has been read, and `'error'` handles any read failures gracefully.
