## **Question 1:** Create a minimal router with two pages: Home (/) and About (/about). Show the links to move between pages.

**Answer:**

```jsx
// App.jsx
import React from 'react';
import { BrowserRouter, Routes, Route, Link } from 'react-router-dom';

function Home() {
  return <h2>Home Page</h2>;
}

function About() {
  return <h2>About Page</h2>;
}

function App() {
  return (
    <BrowserRouter>
      <nav>
        <Link to="/">Home</Link> | <Link to="/about">About</Link>
      </nav>
      <Routes>
        <Route path="/" element={<Home />} />
        <Route path="/about" element={<About />} />
      </Routes>
    </BrowserRouter>
  );
}

export default App;
```

- `Routes`/`Route` map a URL path to the component that should render for it.
- `Link` renders an `<a>` tag that changes the URL **without** a full page reload, unlike a normal `<a href>`.

## **Question 3:** Build a `DashboardLayout` that shows a heading and a menu. Inside it, render `DashboardHome` at `/dashboard` and `Settings` at `/dashboard/settings`.

**Answer:**

```jsx
// DashboardLayout.jsx
import React from 'react';
import { Link, Outlet } from 'react-router-dom';

function DashboardLayout() {
  return (
    <div>
      <h2>Dashboard</h2>
      <nav>
        <Link to="/dashboard">Home</Link> | <Link to="/dashboard/settings">Settings</Link>
      </nav>
      <hr />
      <Outlet />
    </div>
  );
}

export default DashboardLayout;
```

```jsx
// App.jsx
import React from 'react';
import { BrowserRouter, Routes, Route } from 'react-router-dom';
import DashboardLayout from './DashboardLayout';

function DashboardHome() {
  return <p>Welcome to your dashboard!</p>;
}

function Settings() {
  return <p>Dashboard settings go here.</p>;
}

function App() {
  return (
    <BrowserRouter>
      <Routes>
        <Route path="/dashboard" element={<DashboardLayout />}>
          <Route index element={<DashboardHome />} />
          <Route path="settings" element={<Settings />} />
        </Route>
      </Routes>
    </BrowserRouter>
  );
}

export default App;
```

Nested routes let a parent route (`DashboardLayout`) render shared UI (heading, menu), while an `<Outlet />` marks where the matching **child** route's element should appear — `DashboardHome` for the index route, `Settings` for `/dashboard/settings`.

## **Question 4:** Create a route `/users/:userId`. In the `UserPage` component, read the `userId` from the URL and display it.

**Answer:**

```jsx
// UserPage.jsx
import React from 'react';
import { useParams } from 'react-router-dom';

function UserPage() {
  const { userId } = useParams();
  return <h2>Viewing profile for user: {userId}</h2>;
}

export default UserPage;
```

```jsx
// App.jsx
import React from 'react';
import { BrowserRouter, Routes, Route, Link } from 'react-router-dom';
import UserPage from './UserPage';

function UserList() {
  return (
    <nav>
      <Link to="/users/101">User 101</Link> | <Link to="/users/202">User 202</Link>
    </nav>
  );
}

function App() {
  return (
    <BrowserRouter>
      <UserList />
      <Routes>
        <Route path="/users/:userId" element={<UserPage />} />
      </Routes>
    </BrowserRouter>
  );
}

export default App;
```

## **Question 5:** Only logged-in users should see `/dashboard`. Others should be redirected to `/login`. Write a minimal `ProtectedRoute` wrapper.

**Answer:**

```jsx
// ProtectedRoute.jsx
import React from 'react';
import { Navigate } from 'react-router-dom';

function ProtectedRoute({ isLoggedIn, children }) {
  if (!isLoggedIn) {
    return <Navigate to="/login" replace />;
  }
  return children;
}

export default ProtectedRoute;
```

```jsx
// App.jsx
import React, { useState } from 'react';
import { BrowserRouter, Routes, Route, Navigate } from 'react-router-dom';
import ProtectedRoute from './ProtectedRoute';

function Dashboard() {
  return <h2>Welcome to your Dashboard!</h2>;
}

function Login({ onLogin }) {
  return <button onClick={onLogin}>Log in</button>;
}

function App() {
  const [isLoggedIn, setIsLoggedIn] = useState(false);

  return (
    <BrowserRouter>
      <Routes>
        <Route
          path="/dashboard"
          element={
            <ProtectedRoute isLoggedIn={isLoggedIn}>
              <Dashboard />
            </ProtectedRoute>
          }
        />
        <Route path="/login" element={<Login onLogin={() => setIsLoggedIn(true)} />} />
      </Routes>
    </BrowserRouter>
  );
}

export default App;
```

`ProtectedRoute` is a component that checks an `isLoggedIn` condition: if true it renders its `children`; if false it renders a `<Navigate>` element, which immediately redirects to `/login` instead of rendering the protected content.

## **Question 6:** Create a counter slice with increment and decrement. Configure the store and provide it to React.

**Answer:**

```js
// counterSlice.js
import { createSlice } from '@reduxjs/toolkit';

const counterSlice = createSlice({
  name: 'counter',
  initialState: { value: 0 },
  reducers: {
    increment: (state) => {
      state.value += 1;
    },
    decrement: (state) => {
      state.value -= 1;
    },
  },
});

export const { increment, decrement } = counterSlice.actions;
export default counterSlice.reducer;
```

```js
// store.js
import { configureStore } from '@reduxjs/toolkit';
import counterReducer from './counterSlice';

export const store = configureStore({
  reducer: {
    counter: counterReducer,
  },
});
```

```jsx
// index.js
import React from 'react';
import ReactDOM from 'react-dom/client';
import { Provider } from 'react-redux';
import { store } from './store';
import App from './App';

const root = ReactDOM.createRoot(document.getElementById('root'));
root.render(
  <Provider store={store}>
    <App />
  </Provider>
);
```

- `createSlice` generates a reducer plus action creators (`increment`, `decrement`) from a simple object of reducer functions — Redux Toolkit uses Immer internally, so you can write `state.value += 1` as if mutating directly.
- `configureStore` sets up the Redux store (with good defaults like Redux DevTools and thunk middleware already wired in).
- `<Provider store={store}>` makes the store available to any component in the tree via React-Redux's hooks.

## **Question 7:** Build a `Counter` component that shows the value and buttons to increment and decrement using Redux hooks.

**Answer:**

```jsx
// Counter.jsx
import React from 'react';
import { useSelector, useDispatch } from 'react-redux';
import { increment, decrement } from './counterSlice';

function Counter() {
  const value = useSelector((state) => state.counter.value);
  const dispatch = useDispatch();

  return (
    <div>
      <h3>Count: {value}</h3>
      <button onClick={() => dispatch(increment())}>Increment</button>
      <button onClick={() => dispatch(decrement())}>Decrement</button>
    </div>
  );
}

export default Counter;
```

- `useSelector(selectorFn)` subscribes the component to a slice of Redux state, re-rendering it whenever that slice changes.
- `useDispatch()` returns the store's `dispatch` function, used to send actions (like `increment()`/`decrement()`) to the reducer.

## **Question 8:** Create a `fetchProducts` thunk that loads from `/api/products`. Handle loading and error in the slice.

**Answer:**

```js
// productsSlice.js
import { createSlice, createAsyncThunk } from '@reduxjs/toolkit';

export const fetchProducts = createAsyncThunk('products/fetchProducts', async () => {
  const response = await fetch('/api/products');
  if (!response.ok) {
    throw new Error('Failed to fetch products');
  }
  return response.json();
});

const productsSlice = createSlice({
  name: 'products',
  initialState: { items: [], loading: false, error: null },
  reducers: {},
  extraReducers: (builder) => {
    builder
      .addCase(fetchProducts.pending, (state) => {
        state.loading = true;
        state.error = null;
      })
      .addCase(fetchProducts.fulfilled, (state, action) => {
        state.loading = false;
        state.items = action.payload;
      })
      .addCase(fetchProducts.rejected, (state, action) => {
        state.loading = false;
        state.error = action.error.message;
      });
  },
});

export default productsSlice.reducer;
```

```jsx
// ProductList.jsx
import React, { useEffect } from 'react';
import { useSelector, useDispatch } from 'react-redux';
import { fetchProducts } from './productsSlice';

function ProductList() {
  const { items, loading, error } = useSelector((state) => state.products);
  const dispatch = useDispatch();

  useEffect(() => {
    dispatch(fetchProducts());
  }, [dispatch]);

  if (loading) return <p>Loading...</p>;
  if (error) return <p>Error: {error}</p>;

  return (
    <ul>
      {items.map((p) => <li key={p.id}>{p.name}</li>)}
    </ul>
  );
}

export default ProductList;
```

`createAsyncThunk` wraps an async function and automatically dispatches `pending` / `fulfilled` / `rejected` actions around it. The slice's `extraReducers` responds to those three action types to track `loading` and `error` state — showing a spinner while pending, the data once fulfilled, or an error message if rejected.

## **Question 10:** Create a responsive product list: on small screens one column; on medium and up, three columns. Add a hover transition. Also, show how to reuse a utility group with `@apply`.

**Answer:**

```jsx
// ProductList.jsx
import React from 'react';

const products = [
  { id: 1, name: 'Laptop', price: 999 },
  { id: 2, name: 'Phone', price: 599 },
  { id: 3, name: 'Headphones', price: 149 },
];

function ProductList() {
  return (
    <div className="grid grid-cols-1 md:grid-cols-3 gap-4 p-4 bg-gray-100">
      {products.map((p) => (
        <div key={p.id} className="product-card">
          <h3 className="font-bold">{p.name}</h3>
          <p className="text-gray-600">${p.price}</p>
        </div>
      ))}
    </div>
  );
}

export default ProductList;
```

```css
/* index.css - reusing a utility group with @apply */
.product-card {
  @apply bg-white rounded-lg shadow p-4 transition-transform duration-200 hover:scale-105 hover:shadow-lg;
}
```

- `grid grid-cols-1 md:grid-cols-3 gap-4` creates a **1-column grid by default**, switching to **3 columns** at the `md` breakpoint (≥768px) and above.
- `transition-transform hover:scale-105` smoothly scales each card up slightly on hover.
- `@apply` lets you bundle a repeated group of utility classes into a single reusable custom class (`.product-card` above), defined in your CSS file — so you don't have to repeat the same long class list on every card.
