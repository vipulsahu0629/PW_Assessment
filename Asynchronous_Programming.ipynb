## **Question 1:** Write code that creates a new paragraph element with the text **"Welcome to our website!"** and adds it to a div with the id `"content"` on the webpage. Explain what `createElement` and `appendChild` methods do.

### **Answer**

```javascript
// Create a new <p> element
const newParagraph = document.createElement('p');

// Set its text content
newParagraph.textContent = 'Welcome to our website!';

// Select the div with id 'content'
const contentDiv = document.getElementById('content');

// Append the new paragraph as a child of the div
contentDiv.appendChild(newParagraph);
```

**Explanation:**
- `document.createElement(tagName)` creates a brand-new, empty HTML element node in memory
  (e.g. a `<p>` tag). This element does **not** appear on the page until it is inserted
  somewhere in the DOM tree.
- `parentElement.appendChild(childNode)` inserts the given node as the **last child** of
  the parent element, making it visible on the webpage. It physically attaches the newly
  created element to the existing DOM structure.


## **Question 2:** You need to change the appearance of a button when users hover over it. Write code thatfinds a button with id `"myButton"` and changes its background color to blue and text color to white using the `element.style` property.

### **Answer**

```javascript
const myButton = document.getElementById('myButton');

// Change appearance on mouse hover (mouseenter)
myButton.addEventListener('mouseenter', function() {
  myButton.style.backgroundColor = 'blue';
  myButton.style.color = 'white';
});

// Revert appearance when mouse leaves (mouseleave)
myButton.addEventListener('mouseleave', function() {
  myButton.style.backgroundColor = '';
  myButton.style.color = '';
});
```

**Explanation:**
- `document.getElementById('myButton')` selects the button element.
- The `element.style` property lets us directly set **inline CSS styles** via JavaScript
  (e.g. `style.backgroundColor`, `style.color`).
- `mouseenter` fires when the cursor enters the button, and `mouseleave` fires when it
  exits, so the color change is only applied while hovering.


## **Question 3:** Create a button that shows an alert with the message **"Button was clicked!"**. When clicked also, add a second event listener to the same button that changes its text to **"Clicked!"** after being clicked. Explain why `addEventListener` is better than using `onclick`.

### **Answer**

```javascript
const myButton = document.getElementById('myButton');

// First listener: show an alert
myButton.addEventListener('click', function() {
  alert('Button was clicked!');
});

// Second listener: change the button's text
myButton.addEventListener('click', function() {
  myButton.textContent = 'Clicked!';
});
```

**Why `addEventListener` is better than `onclick`:**
- **Multiple handlers:** `addEventListener` allows you to attach *multiple* independent
  event handlers to the same element and event type (as shown above). Using `onclick`
  only allows **one** handler — assigning a second one overwrites the first.
- **Better control:** `addEventListener` supports options like `capture`, `once`, and
  `passive`, and handlers can be removed individually with `removeEventListener`.
- **Separation of concerns:** It keeps JavaScript logic separate from HTML markup
  (unlike inline `onclick="..."` attributes), which is better practice for maintainable code.
- **Works on any target:** It can also listen for events on `window`, `document`, and
  custom events consistently.


## **Question 4:** Write a function named `extractDomain` that takes an email string like`"john@example.com"` and returns the domain part (`"example.com"`) using only string methods. Do not use regular expressions.

### **Answer**

```javascript
function extractDomain(email) {
  // Find the position of the '@' symbol
  const atIndex = email.indexOf('@');

  // Return everything after the '@' symbol
  return email.slice(atIndex + 1);
}

// Example usage
console.log(extractDomain('john@example.com')); // "example.com"
```

**Explanation:**
- `String.prototype.indexOf('@')` returns the index of the first occurrence of `@`.
- `String.prototype.slice(start)` extracts the substring from `start` to the end of the
  string, giving us everything after the `@` symbol — i.e., the domain.


## **Question 5:** Write a function called `repeatMessage` that takes a message string and a number, and returns the message repeated that many times, separated by a space. Do not use loops.
**Example:** `repeatMessage("hello", 3)` should return `"hello hello hello"`

### **Answer**

```javascript
function repeatMessage(message, count) {
  // Base case: no repetitions left
  if (count <= 0) return '';

  // Base case: last repetition (no trailing space)
  if (count === 1) return message;

  // Recursive case: message + space + remaining repetitions
  return message + ' ' + repeatMessage(message, count - 1);
}

// Example usage
console.log(repeatMessage('hello', 3)); // "hello hello hello"
```

**Explanation:**
Since loops (`for`, `while`) are not allowed, the function uses **recursion** instead.
Each call reduces `count` by 1 and appends the message with a space, until `count`
reaches 1 (base case), at which point recursion stops.


## **Question 6:** Explain what the Event Loop is in JavaScript and predict the output of the following code. Also explain why the output appears in that specific order.

```javascript
console.log('Start');

setTimeout(function() {
  console.log('Timeout 1');
}, 0);

console.log('Middle');

setTimeout(function() {
  console.log('Timeout 2');
}, 0);

console.log('End');
```

### **Answer**

**What is the Event Loop?**
JavaScript is single-threaded, meaning it can only execute one piece of code at a time.
The **Event Loop** is the mechanism that allows JavaScript to handle asynchronous
operations (like `setTimeout`, network requests, or events) without blocking the main
thread. It continuously checks the **call stack** (currently executing code) and, once
it's empty, moves callback functions from the **task queue** (also called the callback
queue) into the call stack for execution.

**Predicted Output:**
```
Start
Middle
End
Timeout 1
Timeout 2
```

**Why this order?**
1. All **synchronous** code runs first, in order, because it executes immediately on the
   call stack: `'Start'`, then `'Middle'`, then `'End'`.
2. `setTimeout`, even with a delay of `0`, does **not** run immediately. It hands its
   callback off to the browser/Node's timer facility, and the callback is placed in the
   task queue only after the specified delay.
3. The Event Loop only pushes queued callbacks onto the call stack **after** the current
   synchronous code has fully finished executing (call stack is empty).
4. Since the first `setTimeout` was queued before the second, `'Timeout 1'` logs before
   `'Timeout 2'`.


## **Question 7:** Write a JavaScript function named `startCountdown` that logs "Time left: X" every second, starting from 5 down to 1, and finally logs "Time's up!". Use `setInterval` and `clearInterval` only.

### **Answer**

```javascript
function startCountdown() {
  let timeLeft = 5;

  const intervalId = setInterval(function() {
    if (timeLeft > 0) {
      console.log('Time left: ' + timeLeft);
      timeLeft--;
    } else {
      console.log("Time's up!");
      clearInterval(intervalId); // Stop the interval
    }
  }, 1000);
}

startCountdown();
```

**Explanation:**
- `setInterval(callback, 1000)` repeatedly runs the callback every 1000ms (1 second).
- Each tick checks `timeLeft`; while it's greater than 0, it logs the countdown and
  decrements the counter.
- Once `timeLeft` reaches 0, it logs `"Time's up!"` and calls `clearInterval(intervalId)`
  to stop further executions — otherwise the interval would run forever.


## **Question 8:** Explain what Promises are in JavaScript and write code that creates a Promise to simulate checking if a user is logged in. The Promise should resolve with user data after 2 seconds if successful, or reject with an error message if the user is not found.

### **Answer**

**What is a Promise?**
A `Promise` is a JavaScript object that represents the eventual completion (or failure) of
an asynchronous operation. It has three states:
- **Pending** – initial state, operation not yet complete.
- **Fulfilled** – operation completed successfully (`resolve` was called).
- **Rejected** – operation failed (`reject` was called).

Promises let us write cleaner asynchronous code compared to plain callbacks, and they
can be chained using `.then()` / `.catch()`, or consumed with `async/await`.

```javascript
function checkUserLogin(isUserFound) {
  return new Promise(function(resolve, reject) {
    setTimeout(function() {
      if (isUserFound) {
        resolve({ id: 1, name: 'John Doe', email: 'john@example.com' });
      } else {
        reject('Error: User not found');
      }
    }, 2000); // simulate 2-second delay
  });
}

// Usage
checkUserLogin(true)
  .then(userData => console.log('Login successful:', userData))
  .catch(error => console.error(error));
```

**Explanation:**
- `new Promise((resolve, reject) => {...})` takes an *executor* function with `resolve`
  and `reject` callbacks.
- Inside `setTimeout`, after a 2-second delay, we call `resolve(userData)` if the user is
  found, or `reject(errorMessage)` otherwise.
- The caller consumes the result using `.then()` for success and `.catch()` for failure.


## **Question 9:** Rewrite the following Promise-based code to use `async/await` syntax. The code fetches user data and then fetches that user's posts. Include proper error handling.

```javascript
function getUserDataAndPosts(userId) {
  return fetchUserData(userId)
    .then(userData => {
      return fetchUserPosts(userData.id)
        .then(posts => {
          return { user: userData, posts: posts };
        });
    })
    .catch(error => {
      console.error('Error:', error);
      throw error;
    });
}
```

### **Answer**

```javascript
async function getUserDataAndPosts(userId) {
  try {
    const userData = await fetchUserData(userId);
    const posts = await fetchUserPosts(userData.id);

    return { user: userData, posts: posts };
  } catch (error) {
    console.error('Error:', error);
    throw error;
  }
}
```

**Explanation:**
- Declaring the function with `async` allows the use of `await` inside it, and makes the
  function automatically return a Promise.
- `await fetchUserData(userId)` pauses execution until the Promise resolves, giving us
  the resolved value directly (instead of chaining `.then()`).
- `await fetchUserPosts(userData.id)` works the same way for the second asynchronous call.
- The `try...catch` block replaces `.catch()` — any rejected Promise (from either
  `await` call) throws an error that is caught here, logged, and re-thrown so the caller
  can still handle it.
- This version is much easier to read than nested `.then()` chains, while preserving the
  exact same error-handling behavior as the original.


## **Question 10:** Write code using the fetch API to get weather data from a public API. Display the weather information in a user-friendly format.

```javascript
const apiKey = 'abcd1234';
const apiUrl =
  `https://api.openweathermap.org/data/2.5/weather?q=Mumbai&appid=${apiKey}&units=metric`;
```

### **Answer**

```javascript
const apiKey = 'abcd1234';
const apiUrl =
  `https://api.openweathermap.org/data/2.5/weather?q=Mumbai&appid=${apiKey}&units=metric`;

async function getWeather() {
  try {
    const response = await fetch(apiUrl);

    if (!response.ok) {
      throw new Error(`Weather data not found (status: ${response.status})`);
    }

    const data = await response.json();

    // Display weather information in a friendly format
    console.log(`Weather in ${data.name}:`);
    console.log(`Temperature: ${data.main.temp}°C`);
    console.log(`Feels like: ${data.main.feels_like}°C`);
    console.log(`Condition: ${data.weather[0].description}`);
    console.log(`Humidity: ${data.main.humidity}%`);
    console.log(`Wind Speed: ${data.wind.speed} m/s`);
  } catch (error) {
    console.error('Failed to fetch weather data:', error.message);
  }
}

getWeather();
```

**Explanation:**
- `fetch(apiUrl)` sends an HTTP GET request to the OpenWeatherMap API and returns a
  Promise that resolves to a `Response` object.
- `response.ok` checks whether the HTTP status is in the 200-299 range; if not, an error
  is thrown.
- `response.json()` parses the response body as JSON (also returns a Promise, hence the
  second `await`).
- The relevant fields (`name`, `main.temp`, `main.feels_like`, `weather[0].description`,
  `main.humidity`, `wind.speed`) are extracted from the JSON and logged in a clean,
  readable format.
- The `try...catch` block handles network failures or bad responses gracefully.
